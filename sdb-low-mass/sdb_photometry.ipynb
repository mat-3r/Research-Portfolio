**Introduction/Motivation**

A binary star system consists of two stars that are gravitationally bound together and orbits closely with another star. These systems are important for understanding stellar evolution and how the two components affect each other's evolution. GALEX J082053.6+000843 is a perfect example of such system, as it is an eclipsing subdwarf B (sdB) star binary accompanied by a brown dwarf companion. The primary star is a compact, very hot subdwarf star located on the extreme horizontal branch of stellar evolution. These stars are core helium burning stars with hydrogen envelopes too thin to keep up with hydrogen shell burning. (Geier et al. 2011) Its companion is a tidally locked M type brown dwarf that orbits the primary star. Due to tidal interaction, it constantly only shows one side of its sphere to its primary star. This indicates strong gravitational interactions between the two bodies as they are so close together. 
This system offers a unique opportunity to conduct the effects of strong irradiation and the orbital evolution of binaries with low mass companions. So by analyzing the lights curve, which shows how the system's brightness changes over time, we can obtain valuable knowledge into the system's orbit, structure, influence of reflection effects, and interaction between the two stars. The ultimate motivation is to analyze 269 science exposures using AstroImageJ to compute the light curve, phase folding, and determine the orbital period using Lomb-Scargle. This analysis will be compared with the results obtained from the MUSCHFUSS project and theoretical expectations. MUSHFUSS project primary objective was to find sdBs with compact companions such as white dwarfs, black holes or neutron stars. (Geier et al. 2011) Analyzing the photometric light curves of the system, we can obtain the orbital period and understand its reflection effects. Ultimately, this analysis will help us uncover the more complex ideas pertaining to the end stages of binary evolution.

**Observations/Theoretical Expectations** 

The photometric behaviors I observed for GALEX J082053.6+000843 are consistent with the theoretical expectations for the light curve of a sdB+low-mass brown dwarf binary system. The main eclipse displays a strong dip in relative flux over time, connecting to the companion star transiting in front of the sdB star, blocking it from view, and the cooler side of brown dwarf is facing us. As the companion orbits orbits around the main star, the system becomes the brightest around orbital phase 0.24. Its like about 135 degrees into its orbit which is when the heated side of the companion starts to face us. This produces reflection effect which we are able to observe. The light curve produced by AstroImageJ is consistent with the light curve produced with python, advocating the accuracy of the results further. Since the system is tidally locked, it is expected to show irradiation effects which the observed brightness changes support  this idea. The eclipse phases further confirm the role of reflection effects in changing the observed flux throughout the orbit. The observation revealed three phases in the phase folded light curve, which supports the hypothesis of a reflection effect causing the light to appear brighter in some regions. The three phases were the primary eclipse, the companion irradiated flux, and the companion star. The first bump is the brown dwarf transitioning behind the sdB star. This causes a slow decline which is the irradiated side of the brown dwarf turning away. The third bump is the cooler side of the brown dwarf moving in front of the sdB star. This accurately represents the overall light curve shape which aligns with the theoretical models for expected sdB and brown dwarf systems. The magnitude was observed to increase during these orbital phases which is consistent with the heating of the brown dwarf's side and its contribution of reflected light to the entire system. This was concurrent with the theoretical expectations of the brown dwarf being able to influence later stages of stellar evolution in close binary systems.
